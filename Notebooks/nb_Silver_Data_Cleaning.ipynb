**Numeric Duration Validation**

```python
for col in ['Queue Wait Hours','User Info Wait Hours',
            'Grc Approval Wait Hours','Fulfilment Hours',
            'Closure Lag Hours','Total Resolution Hours']:
    count_neg = (df[col] < 0.0).sum()
    if count_neg > 0:
        print(f"{col}: {count_neg} negative values")
else:
        print("No Negetives")
```

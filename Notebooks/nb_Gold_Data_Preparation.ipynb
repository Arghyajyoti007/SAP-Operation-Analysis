```python
# Adding date key in fact_operational_request (Enriched with Date Keys)
df_operational_request = pd.read_parquet("abfss://ws_SAP_GRC_Operation@onelake.dfs.fabric.microsoft.com/lh_SAP_GRC_Operation.Lakehouse/Tables/Silver/silver_operational_request")
df_operational_request['Created_Date_Key'] = df_operational_request['Request_Created_Datetime'].dt.strftime('%Y%m%d').astype(int)
df_operational_request['Closure_Date_Key'] = df_operational_request['Closure_Datetime'].dt.strftime('%Y%m%d').astype('Int64')
```

```python
df_operational_request['Request_Status'].unique()
```

```python
# Adding Days Column in fact_operational_request (Days columns)
df_operational_request['Total_Resolution_Days'] = (df_operational_request['Total_Resolution_Hours'] / 24.0).round(2)
df_operational_request['Grc_Approval_Wait_Days'] = (df_operational_request['Grc_Approval_Wait_Hours'] / 24.0).round(2)
df_operational_request['User_Info_Wait_Days'] = (df_operational_request['User_Info_Wait_Hours'] / 24.0).round(2)
df_operational_request['SLA_Target_Days'] = (df_operational_request['SLA_Target_Hours'] / 24.0).round(1)
```

```python
df_operational_request.info()
```

```python
# Save to Gold Lakehouse via Spark
spark_gold_fact = spark.createDataFrame(df_operational_request)
spark_gold_fact.write \
    .format("delta") \
    .mode("overwrite") \
    .save("Tables/Gold/fact_operational_request")
```
